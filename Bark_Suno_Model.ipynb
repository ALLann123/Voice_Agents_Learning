**================Bark-Suno==================**

What is Bark?
- Is an AI text to audio generation model developed by Suno.
- Unlike tradional text-to-speech systems that primarily map:
	text--> Speech
- Bark is a generative audio model. It can generate speech while also modeling things
such as:
	1. Pauses
	2. tone
	3. emotion
	4. laughter
	5. non-speech sounds
	6. multilingual speech
	7. expressive delivery

**====2. What does "Text-to-Speech" actually mean?====**
- Work flow:
***********
"I am Qwerty"

       ↓

Speech representation

       ↓
Audio waveform
**************
- Converts our text to a waveform representation of the sound
- Computer speakers convert that waveform into vibrations, and those vibrations become
what we hear as speech.

```python
from transformers import AutoProcessor, BarkModel

processor = AutoProcessor.from_pretrained("suno/bark")
model = BarkModel.from_pretrained("suno/bark")
```

```python
import torch

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    gpu = torch.cuda.get_device_name(0)
    memory = torch.cuda.get_device_properties(0).total_memory / 1024**3

    print("GPU:", gpu)
    print(f"GPU memory: {memory:.2f} GB")
else:
    print("No GPU detected")
```

```python
import torch
from IPython.display import Audio, display

voice_preset="v2/en_speaker_6"

def to_audio(text):
  """Will generate the audio here on collab"""
  # converts text into the representation Bark expects
  inputs=processor(
      text,
      voice_preset=voice_preset,
  )

  # actual generation
  with torch.inference_mode():
    audio_array=model.generate(**inputs)

  audio_array=audio_array.cpu().numpy().squeeze()

  print("Audio Generated!")
  print("Audio Samples: ", len(audio_array))

  # lets colab play the generated waveform directly
  display(Audio(audio_array, rate=model.generation_config.sample_rate))

```

```python
# test our function
text="Hello, I am Fred a local AI virtual Assistant."

to_audio(text)
```

```python
# Move it to GPU
device = "cuda" if torch.cuda.is_available() else "cpu"

model = model.to(device)

print("Model device:", next(model.parameters()).device)
```

```python
def to_audio(text):
    """Generate Bark audio on the Colab GPU."""

    # Prepare text + speaker preset
    inputs = processor(
        text,
        voice_preset=voice_preset,
    )

    # Move ALL tensors, including nested history_prompt,
    # to the GPU.
    inputs = inputs.to(device)

    print("Input device:", inputs["input_ids"].device)

    if "history_prompt" in inputs:
        print("Voice preset loaded.")

        for key, value in inputs["history_prompt"].items():
            print(f"{key}: {value.device}")

    # Generate audio
    with torch.inference_mode():
        audio_array = model.generate(
            **inputs,
            semantic_max_new_tokens=768 # the amount of audio time
        )

    # Move generated audio back to CPU
    audio_array = audio_array.cpu().numpy().squeeze()

    print("Audio Generated!")
    print("Audio Samples:", len(audio_array))

    display(
        Audio(
            audio_array,
            rate=model.generation_config.sample_rate
        )
    )
```
